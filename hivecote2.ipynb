# HIVE-COTE 2.0

Usage example for the HIVE-COTE 2.0 (HC2) classifier from the paper "HIVE-COTE 2.0: a new meta ensemble for time series classification" published in Machine Learning. See the [README](README.md) for the results from the paper.

HC2 is available in the [aeon](https://github.com/aeon-toolkit/aeon) toolkit. It follows the `scikit-learn` interface and handles univariate and multivariate series.

```python
import warnings

from aeon.classification.hybrid import HIVECOTEV2
from aeon.datasets import load_basic_motions, load_gunpoint
from sklearn.metrics import accuracy_score

warnings.filterwarnings("ignore")
```

## Univariate data

GunPoint is one of the 112 UCR datasets used in the paper.

```python
X_train, y_train = load_gunpoint(split="train")
X_test, y_test = load_gunpoint(split="test")

hc2 = HIVECOTEV2(random_state=0)
hc2.fit(X_train, y_train)

preds = hc2.predict(X_test)
print(f"Accuracy: {accuracy_score(y_test, preds):.4f}")
```

HC2 can take a long time to train on larger datasets. Setting a time limit contracts the whole ensemble, splitting the time between its four components.

```python
hc2 = HIVECOTEV2(time_limit_in_minutes=1, random_state=0)
hc2.fit(X_train, y_train)

preds = hc2.predict(X_test)
print(f"Accuracy: {accuracy_score(y_test, preds):.4f}")
```

## Multivariate data

BasicMotions is one of the 26 UEA datasets used in the paper.

```python
X_train, y_train = load_basic_motions(split="train")
X_test, y_test = load_basic_motions(split="test")

hc2 = HIVECOTEV2(random_state=0)
hc2.fit(X_train, y_train)

preds = hc2.predict(X_test)
print(f"Accuracy: {accuracy_score(y_test, preds):.4f}")
```
